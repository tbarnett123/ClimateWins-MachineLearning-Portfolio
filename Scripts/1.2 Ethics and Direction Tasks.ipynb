```python
#importing dataframes 
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import sklearn
from sklearn.preprocessing import StandardScaler
```

```python
#importing Data Set
df = pd.read_csv(os.path.join(path, "Dataset-weather-prediction-dataset-processed.csv"))
```

```python
df.head()
```

```python
df.shape
```

```python
#checking missing values
df.isnull().sum().sum()
```

```python
#excluding these columns from scaling
exclude_cols = ['DATE', 'MONTH',]
exclude_cols = [c for c in exclude_cols if c in df.columns]
exclude_cols
```

```python
#builing the ones that will be scaled
scale_cols = [c for c in df.columns if c not in exclude_cols]
len(scale_cols), scale_cols[:10]
```

```python
#creating the scaler
scaler = StandardScaler()
```

```python
#scaling columns
df_scaled = df.copy()
df_scaled[scale_cols] = scaler.fit_transform(df[scale_cols])
df_scaled.head()
```

```python
#Checking values of the mean for the scaled variables
df_scaled[scale_cols].mean().head().round(4)
```

```python
#verifying scaling by checking the standard deviation of scaled variables
df_scaled[scale_cols].std(ddof=0).head().round(4)
```

```python
#Exporting scaled dataframe 
output_name = "Dataset-weather-prediction-dataset-processed_scaled.csv"
df_scaled.to_csv(os.path.join(path, output_name), index=False)

os.path.join(path, output_name)
```
